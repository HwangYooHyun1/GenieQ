```python
import json

def filter_unique_data(json_file, output_file):
    with open(json_file, "r", encoding="utf-8") as file:
        data = json.load(file)
    
    # passage가 동일한 경우 하나의 데이터만 유지하기 위한 딕셔너리
    unique_data = {}
    
    for item in data:
        passage = item.get("passage", "").strip()
        subject = item.get("subject", "").strip()
        topics = list(set(item.get("topics", [])))  # 중복 제거
        
        if passage not in unique_data:
            unique_data[passage] = {
                "passage": passage,
                "subject": subject,
                "topics": topics
            }
    
    # 결과를 새로운 JSON 파일로 저장
    with open(output_file, "w", encoding="utf-8") as out_file:
        json.dump(list(unique_data.values()), out_file, ensure_ascii=False, indent=4)
    
    print(f"Filtered data saved to {output_file}")
```

```python
filter_unique_data("./data_preprocessed/all_data_preprocessed.json", "./data_preprocessed/only_passage.json")
```
